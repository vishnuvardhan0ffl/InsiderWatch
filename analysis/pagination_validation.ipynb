```python
from collectors.polymarket import (
    TradeQuery,
    collect_trades,
    save_trade_collection
)
```

```python
MARKET_ID = "0xd1e4e03a0129aad7b23e835bfb5c0166d7342fb98e91aa7dd0f3e6cb423d9c21"

print(MARKET_ID)
```

```python
from datetime import datetime, timezone

start = int(
    datetime(
        2025,
        1,
        1,
        tzinfo=timezone.utc
    ).timestamp()
)

end = int(
    datetime(
        2026,
        4,
        28,
        10,
        59,
        59,
        tzinfo=timezone.utc
    ).timestamp()
)

print(start)
print(end)
```

```python
query = TradeQuery(
    market=MARKET_ID,
    start=start,
    end=end,

    # Important:
    # explicitly include maker + taker records
    taker_only=False
)
```

```python
trades, metadata = collect_trades(
    query,
    window_s=7 * 86400
)
```

```python
print(
    "Total trades:",
    len(trades)
)
```

```python
import pandas as pd

df = pd.DataFrame(
    trades
)

print(
    "Rows:",
    len(df)
)
```

```python
duplicate_rows = (
    df.astype(str)
      .duplicated()
      .sum()
)

print(
    "Duplicate rows:",
    duplicate_rows
)
```
